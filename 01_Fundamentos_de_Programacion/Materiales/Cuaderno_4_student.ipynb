## 2 · Modulos

Un **modulo** es un archivo `.py` con funciones, variables y clases.
Permite organizar y reutilizar codigo entre proyectos.

```python
# Formas de importar:
import math                     # importar el modulo completo
from math import sqrt, pi       # importar funciones especificas
from math import *              # importar todo (evitar en proyectos grandes)
import numpy as np              # alias (muy comun)
import os.path as osp           # modulo con submódulo
```

### Modulos de la biblioteca estandar mas utiles

| Modulo | Descripcion |
|--------|-------------|
| `math` | Funciones matematicas |
| `random` | Numeros aleatorios |
| `datetime` | Fechas y tiempos |
| `os` | Interaccion con el sistema operativo |
| `os.path` | Manipulacion de rutas de archivos |
| `csv` | Lectura/escritura de archivos CSV |
| `json` | Serializacion JSON |
| `statistics` | Estadisticas basicas |
| `collections` | Estructuras de datos avanzadas |

````python
import statistics
from datetime import date, timedelta

# statistics: media, desviacion estandar, mediana
rendimientos = [8.2, 5.1, 9.4, 3.7, 11.2, 6.8, 7.5, 4.9]

print(f"Media:      {statistics.mean(rendimientos):.2f}")
print(f"Mediana:    {statistics.median(rendimientos):.2f}")
print(f"Desv. est.: {statistics.stdev(rendimientos):.2f}")
print(f"Varianza:   {statistics.variance(rendimientos):.2f}")

# datetime: calcular dias entre fechas
siembra  = date(2024, 3, 15)
cosecha  = date(2024, 8, 20)
ciclo    = cosecha - siembra
print(f"\nDuracion del ciclo: {ciclo.days} dias")

# Calcular fecha de proxima aplicacion
hoy       = date.today()
proxima   = hoy + timedelta(days=14)
print(f"Proxima aplicacion: {proxima.strftime('%d/%m/%Y')}")
````

## 3 · Archivos de texto

Python puede leer y escribir archivos con la funcion `open()`.
El bloque `with` garantiza que el archivo se cierre correctamente.

```python
with open("ruta/archivo.txt", "modo", encoding="utf-8") as f:
    # operaciones con f
```

| Modo | Descripcion |
|------|-------------|
| `"r"` | Solo lectura (default) |
| `"w"` | Escritura — sobreescribe si existe |
| `"a"` | Agregar al final (append) |
| `"r+"` | Lectura y escritura |

**Metodos principales:**
- `f.read()` — todo el contenido como una cadena
- `f.readline()` — una linea
- `f.readlines()` — lista de lineas
- `f.write(texto)` — escribir texto
- `f.writelines(lista)` — escribir lista de cadenas
