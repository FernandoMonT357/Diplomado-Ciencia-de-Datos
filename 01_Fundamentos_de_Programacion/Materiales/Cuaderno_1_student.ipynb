## 2 · Jupyter Notebook y Google Colab

### Los cuadernos de Jupyter Notebook

En los cuadernos de **Jupyter Notebook** trabajamos con dos tipos principales de celdas:

- 📝 **Celdas de Markdown**  
  Se usan para escribir texto con formato. Permiten agregar títulos, listas, enlaces, imágenes, ecuaciones matemáticas y más. Son útiles para documentar el código y explicar lo que estamos haciendo.

- 💻 **Celdas de código**  
  Aquí se escribe y ejecuta código, como Python. Al ejecutar una celda, el resultado aparece justo debajo. Esto permite probar el código paso a paso y ver sus efectos en tiempo real.

#### ¿Por qué usar un cuaderno de Jupyter en lugar de un script?

Aunque los scripts tradicionales (como archivos `.py` en Python) son muy útiles para desarrollar programas completos, los cuadernos de **Jupyter Notebook** ofrecen ventajas cuando se trata de _aprender, explorar datos o analizar paso a paso_.

##### Interactividad

Se puede ejecutar código **celda por celda**, lo que facilita probar fragmentos de código y ver los resultados de inmediato sin tener que correr todo el programa.

##### Documentación integrada

Permite combinar **código con texto explicativo** usando Markdown, además de insertar imágenes, fórmulas matemáticas, enlaces, y más. Esto mejora la claridad y hace el análisis mucho más comprensible.


##### Visualización directa

Los **gráficos y tablas** generados por bibliotecas como `matplotlib`, `seaborn` o `pandas` se muestran justo debajo del código que los crea, lo que permite una interpretación más rápida y ordenada.

##### Reproducibilidad

Es ideal para informes y análisis donde se quiere mostrar **paso a paso** qué se hizo y qué resultados se obtuvieron. Todo queda registrado en un mismo lugar.


##### Ideal para enseñanza y aprendizaje

Facilita el proceso educativo al permitir **explicar conceptos** junto con su implementación en código, favoreciendo una comprensión más profunda.

## 3 · Variables y tipos de datos

### 3.1 Variables

Una **variable** es un espacio en memoria que almacena datos que pueden cambiar durante la ejecución del programa.

**Características:**
- Tienen un **nombre** (identificador)
- Almacenan un **valor**
- Tienen un **tipo de dato**

```python
rendimiento_ton_ha = 8.2   # la etiqueta "rendimiento_ton_ha" guarda 8.2
```

### Reglas de nombramiento — PEP 8

| Estilo | Uso |
|--------|-----|
| `snake_case` | Variables y funciones: `area_parcela`, `fecha_siembra` |
| `MAYUSCULAS` | Constantes: `TASA_IVA`, `MAX_HUMEDAD` |
| `PascalCase` | Clases (mas adelante en el diplomado) |

**Prohibido:**
- Comenzar con numero: ~~`2parcela`~~
- Espacios: ~~`area parcela`~~
- Palabras reservadas: ~~`if`~~, ~~`for`~~, ~~`class`~~

**Balance entre claridad y brevedad**:
  
❌ `numero_de_estudiantes_matriculados_en_el_curso`

### 3.2 Tipos de datos

| Tipo de Dato      | Categoría       | Mutable | Ejemplo                  | Descripción                              |

| `int`            | Numérico        | ❌      | `42`                     | Enteros sin límite de tamaño             |

- **Ordenado**: Mantiene el orden de inserción (list, tuple, str)

- **Indexable**: Se puede acceder por índice (list, tuple, str)

### 4.4 Operadores de asignacion compuesta

Atajos para actualizar el valor de una variable:

```python
x += 5   # equivale a  x = x + 5
x -= 3   # equivale a  x = x - 3
x *= 2   # equivale a  x = x * 2
x /= 4   # equivale a  x = x / 4
```

````python
# Acumular cosecha durante varios dias
cosecha_total = 0

cosecha_total += 1250   # dia 1 @blank
cosecha_total += 1380   # dia 2 @blank
cosecha_total += 980    # dia 3 @blank
cosecha_total += 1450   # dia 4 @blank
cosecha_total += 1320   # dia 5 @blank

print(f"Cosecha acumulada: {cosecha_total} kg")
print(f"Equivale a:        {cosecha_total / 1000:.2f} toneladas")

# Aplicar descuento al precio
precio = 4500.0
precio *= 0.95   # descuento del 5%
print(f"Precio con descuento: ${precio:,.2f}")
````

## 5 · Salida de datos: print() y f-strings

Las **f-strings** son la forma moderna y preferida de formatear texto en Python.
Se escriben con una `f` antes de las comillas y permiten insertar variables directamente.

```python
nombre = "Lote Norte"
area   = 25.5
print(f"El {nombre} tiene {area} ha")
```

### Especificadores de formato dentro de `{}`

| Especificador | Resultado | Ejemplo |
|--------------|-----------|---------|
| `:.2f` | 2 decimales | `f"{8.2345:.2f}"` → `8.23` |
| `:,` | separador de miles | `f"{12500:,}"` → `12,500` |
| `:>10` | alineado a la derecha en 10 caracteres | |
| `:<10` | alineado a la izquierda | |
| `:.1%` | porcentaje con 1 decimal | `f"{0.856:.1%}"` → `85.6%` |
