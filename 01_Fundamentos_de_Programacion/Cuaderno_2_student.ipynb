### 3.4 Formateo avanzado con f-strings

Los especificadores de formato dentro de `{}` permiten controlar la presentacion:

```python
f"{valor:.2f}"    # 2 decimales
f"{valor:,}"      # separador de miles
f"{valor:>10}"    # alinear a la derecha en 10 caracteres
f"{valor:<10}"    # alinear a la izquierda
f"{valor:^10}"    # centrar
f"{valor:010}"    # rellenar con ceros
f"{valor:.1%}"    # porcentaje con 1 decimal
f"{valor:e}"      # notacion cientifica
```

## 4 · Expresiones regulares (modulo re)

Las expresiones regulares (regex) son patrones para buscar, extraer
y transformar texto de forma avanzada.

```python
import re
```

| Funcion | Descripcion |
|---------|-------------|
| `re.search(patron, texto)` | Primera coincidencia en el texto |
| `re.findall(patron, texto)` | Lista de todas las coincidencias |
| `re.sub(patron, reemplazo, texto)` | Reemplazar coincidencias |
| `re.split(patron, texto)` | Dividir por patron |
| `re.match(patron, texto)` | Coincidencia al inicio del texto |

### Metacaracteres basicos

| Patron | Descripcion | Ejemplo |
|--------|-------------|---------|
| `\d` | Digito [0-9] | `\d+` = uno o mas digitos |
| `\w` | Caracter de palabra | `\w+` = una palabra |
| `\s` | Espacio en blanco | |
| `.` | Cualquier caracter | |
| `+` | Uno o mas | |
| `*` | Cero o mas | |
| `?` | Cero o uno | |
| `^` | Inicio de cadena | |
| `$` | Fin de cadena | |
| `[abc]` | a, b o c | |
