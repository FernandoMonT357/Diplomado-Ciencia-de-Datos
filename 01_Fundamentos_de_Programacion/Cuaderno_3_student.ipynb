El bloque indentado se ejecuta solo si la condición es verdadera.

La condición se evalúa siempre

- Si el resultado es True se ejecuta el bloque de sentencias

### 4.1 Expresion ternaria (if en una linea)

Para asignaciones simples puedes usar la expresion ternaria:
```python
valor = x if condicion else y
```

````python
rendimientos = [8.2, 5.1, 9.4, 3.7, 11.2]

salida = "\n".join(
    f"  {r:.1f} ton/ha  [{'Bien' if r >= 8 else 'Mal'}] {'Alto' if r >= 8 else 'Bajo'}"
    for r in rendimientos
)

print(salida)
````
